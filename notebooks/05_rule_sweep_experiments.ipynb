# 05 — Rule sweep experiments

MSE vs number of rules (1–50) for first-order ANFIS and IT2-ANFIS, 10 seeds per rule count; plot both side by side.

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
sys.path.insert(0, str(Path.cwd().parent))

from src.data import load_csv, clean_and_prepare
from src.features import select_features_mi
from src.experiments.sweep import sweep_rules

path = Path("../data/raw/Data-Melbourne_F_fixed.csv")
if not path.exists():
    path = Path("../../Data-Melbourne_F_fixed.csv")
df = load_csv(path)
df_clean = clean_and_prepare(df)
df_final = select_features_mi(df_clean)
X = df_final.drop(columns=['Energy Consumption']).values
y = df_final['Energy Consumption'].values
```

```python
sweep_first = sweep_rules(X, y, model_type='anfis_first', n_rules_min=1, n_rules_max=25, n_seeds=5)
sweep_it2 = sweep_rules(X, y, model_type='it2', n_rules_min=1, n_rules_max=25, n_seeds=5)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
ax1, ax2 = axes
n1 = sweep_first['n_rules_list']
m1 = sweep_first['mse_mean']
s1 = sweep_first['mse_std']
ax1.errorbar(n1, m1, yerr=s1, fmt='o-', capsize=3)
ax1.set_xlabel('Number of rules')
ax1.set_ylabel('Test MSE')
ax1.set_title('First-order ANFIS')
ax1.grid(True, alpha=0.3)
n2 = sweep_it2['n_rules_list']
m2 = sweep_it2['mse_mean']
s2 = sweep_it2['mse_std']
ax2.errorbar(n2, m2, yerr=s2, fmt='s-', capsize=3, color='green')
ax2.set_xlabel('Number of rules')
ax2.set_ylabel('Test MSE')
ax2.set_title('IT2-ANFIS')
ax2.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
