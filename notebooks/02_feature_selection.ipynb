# 02 — Feature selection

Mutual information with target, visualize selected features, show final df head, and plot target density.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from src.data import load_csv, clean_and_prepare
from src.features import select_features_mi, compute_mi_scores
from src.viz import plot_mi_scores, plot_target_density
from src.config import TARGET_COLUMN, RAW_DIR, DEFAULT_CSV_NAME
import matplotlib.pyplot as plt

path = RAW_DIR / DEFAULT_CSV_NAME
if not path.exists():
    path = Path("../data/raw/Data-Melbourne_F_fixed.csv")
df = load_csv(path)
df_clean = clean_and_prepare(df)
X = df_clean.drop(columns=[TARGET_COLUMN])
y = df_clean[TARGET_COLUMN]
mi_series = compute_mi_scores(X, y)
plot_mi_scores(mi_series, title="Feature Importance (MI) with Energy Consumption")
plt.show()
```

```python
import matplotlib.pyplot as plt
df_final = select_features_mi(df_clean)
print("Shape after MI selection:", df_final.shape)
df_final.head()
```

```python
plot_target_density(df_final[TARGET_COLUMN].values)
plt.show()
```
